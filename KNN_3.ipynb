### Q1. Write a Python code to implement the KNN classifier algorithm on load_iris dataset in sklearn.datasets.

```python
import pandas as pd 
import numpy as np
import seaborn as sns 
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder, OneHotEncoder
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import r2_score, accuracy_score, recall_score, precision_score, classification_report, confusion_matrix
from sklearn.datasets import load_iris
```

```python
data = load_iris()
data.items()
```

```python
data.DESCR
```

```python
df = pd.DataFrame(data.data, columns= data.feature_names)
df
```

```python
y = data.target
y
```

```python
data.target_names
```

```python
X_train, X_test, y_train, y_test = train_test_split(df, y, test_size=0.20, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
model = KNeighborsClassifier(n_neighbors=5,weights="uniform")
model
```

```python
model.fit(X_train, y_train)
y_predict = model.predict(X_test)
accuracy_score_model = accuracy_score(y_test, y_predict)
print(f"the test accuracy of the model is : {accuracy_score_model}")
```

```python
y_predict_train = model.predict(X_train)
accuracy_score_model_traindata = accuracy_score(y_train, y_predict_train)
print(f"the test accuracy of the model is : {accuracy_score_model_traindata}")
```

#### Q6. Implement a function to standardise the features before applying KNN classifier. 

```python
def stander_scaler(data):
    columns = data.columns

    for i in columns :
        data[i] = (data[i].mean() - data[i])/data[i].std()
    return data 
```

```python
data = load_iris()
df = pd.DataFrame(data.data, columns= data.feature_names)
y = data.target
X_train, X_test, y_train, y_test = train_test_split(df, y, test_size=0.20, random_state=42)
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
y_train
```

```python
X_train
```

```python
X_train = stander_scaler(X_train)
X_train
```

```python
X_test = stander_scaler(X_test)
X_test
```

```python
model = KNeighborsClassifier(n_neighbors=3, weights="distance")
model.fit(X_train,y_train)
y_pre = model.predict(X_test)
y_pre_train = model.predict(X_train)
accuracy_train = accuracy_score(y_train, y_pre_train)
accuracy_test = accuracy_score(y_test, y_pre)

print(f"the train score of the model is {accuracy_train}")
print(f"the test score of the model is {accuracy_test}")
```
